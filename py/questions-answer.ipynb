## 2. is (identity operator)

## why python strings are immutable, gimme answer with exaple 
Concept: Strings are Immutable

Immutable means:

Once a string is created in memory, it cannot be changed.
Any “change” creates a new string object instead of modifying the original.

🧠 Why are Python strings immutable?

Here are four main reasons, explained simply:

1. Memory efficiency & caching

Python reuses string objects (especially small or common strings) for speed.
If strings were mutable, caching and reusing them safely would be impossible.

Example:

a = "hello"
b = "hello"
print(a is b)  # True — both point to same memory (cached)


✅ Because strings are immutable, Python knows a and b will never change, so it safely reuses one copy in memory.

If strings were mutable, changing a would also change b — which would be dangerous!

2. Thread safety

Immutable objects can be shared across multiple threads safely — no accidental overwriting.

Imagine two threads both using the same string "OK".
If strings were mutable, one thread could change "OK" to "NO", breaking the other thread’s logic.

3. Hashing and dictionary keys

Strings are often used as keys in dictionaries or elements in sets.
For an object to be used as a key, its hash must never change.

Since immutability guarantees a string’s value (and therefore its hash) doesn’t change:

user = {"name": "Alice"}
print(user["name"])  # Works fine


If "name" could change its characters, Python would lose track of the key — your dictionary would break.

4. Predictability and simplicity

Every string operation that “modifies” a string actually creates a new one, keeping behavior consistent and predictable.

Example:

text = "hello"
new_text = text.upper()

print(text)      # hello  (original string unchanged)
print(new_text)  # HELLO  (new string created)

🧪 Let’s visualize it:
a = "hi"
print(id(a))   # e.g. 13985000 (memory address)

a += "!"
print(a)       # hi!
print(id(a))   # e.g. 13985200 (different address)


➡️ When you modify a, Python creates a new string object instead of changing the original one.

🧍‍♂️Real-life analogy:

Think of strings like sealed envelopes 📨:
If you want to change the message, you don’t open and edit the old letter —
you write a new letter and put it in a new envelope.

✅ Summary Table
Reason	Explanation
Memory optimization	Reuses strings safely
Thread safety	No shared-state issues
Hashing	Enables use as dictionary keys
Predictability	Every modification returns a new object

```python
#example 
text = "hello"
text2 = text.upper()
print(text2)
print(text)  # Original string remains unchanged
#addresss
print(id(text))
print(id(text2))
```
